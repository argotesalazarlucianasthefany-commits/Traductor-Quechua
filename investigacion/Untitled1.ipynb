```python
import os
import json

import httpx
from dotenv import load_dotenv

load_dotenv()

BASE_URL = "https://openrouter.ai/api/v1"
MODELO_FREE = "nvidia/nemotron-3-ultra-550b-a55b:free"

API_KEY = os.environ.get("OPENROUTER_API_KEY")
assert API_KEY, "Falta OPENROUTER_API_KEY. Copia .env.example a .env y pon tu key ahí."
print("API key cargada correctamente")
```

```python
def preguntar(messages: list[dict], modelo: str = MODELO_FREE) -> str:
    """Envía la lista de mensajes y devuelve el texto de la respuesta."""
    r = httpx.post(
        f"{BASE_URL}/chat/completions",
        headers={"Authorization": f"Bearer {API_KEY}"},
        json={"model": modelo, "messages": messages},
        timeout=60,
    )
    r.raise_for_status()
    return r.json()["choices"][0]["message"]["content"]
```

```python
def preguntar_con_la_web(messages: list[dict], modelo: str = MODELO_FREE) -> str:
    """Envía la lista de mensajes y devuelve el texto de la respuesta."""
    plugins = [
      {
        "id": "web",
        "max_results": 3,
      }
    ]
    r = httpx.post(
        f"{BASE_URL}/chat/completions",
        headers={"Authorization": f"Bearer {API_KEY}"},
        json={"model": modelo, "messages": messages, "plugins": plugins,},
        timeout=60,
    )
    r.raise_for_status()
    return r.json()["choices"][0]["message"]["content"]
```

```python
# 1ª llamada: system + user
messages = [
    {"role": "system", "content": "Eres un profesor de quechua experto."},
    {"role": "user", "content": "¿que signifca tengan un buen dia en quechua?"},
]

respuesta = preguntar_con_la_web(messages)
print("bot>", respuesta)

# Guardamos la respuesta del asistente
messages.append({"role": "assistant", "content": respuesta})
```

```python
# 1ª llamada: system + user
messages = [
    {"role": "system", "content": "Eres un traductor de quechua sureño"},
    {"role": "user", "content": "¿como se dice en tengan buen dia en quechua"},
]

respuesta = preguntar(messages)
print("bot>", respuesta)

# Guardamos la respuesta del asistente
messages.append({"role": "assistant", "content": respuesta})
```

```python
# 1ª llamada: system + user
messages = [
    {"role": "system", "content": "Eres un traductor de quechua de Bolivia q responde de forma breve"},
    {"role": "user", "content": "¿traduce la frase , que estas haciendo?"},
]

respuesta = preguntar_con_la_web(messages)
print("bot>", respuesta)

# Guardamos la respuesta del asistente
messages.append({"role": "assistant", "content": respuesta})
```
